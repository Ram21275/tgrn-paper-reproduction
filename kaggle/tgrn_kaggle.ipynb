# TgRN on Kaggle
Train the independent paper reproduction on a prepared CMU-MOSI or CH-SIMS feature pickle. Enable a GPU and attach the feature dataset before running.

```python
from pathlib import Path
import os, sys, subprocess

REPO = Path('/kaggle/working/tgrn-paper-reproduction')
if not REPO.exists():
    subprocess.run(['git', 'clone', 'https://github.com/ram21275/tgrn-paper-reproduction.git', str(REPO)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-e', str(REPO)], check=True)
os.chdir(REPO)
```

```python
# MOSI feature dataset attached to this Kaggle notebook.
DATA_PATH = Path('/kaggle/input/datasets/ramdabas/ann-proj/unaligned_50.pkl')
CONFIG = Path('configs/mosi.json')  # use configs/sims.json for CH-SIMS
OUTPUT = Path('/kaggle/working/tgrn-results')
assert DATA_PATH.exists(), f'Update DATA_PATH; not found: {DATA_PATH}'
```

```python
subprocess.run([sys.executable, 'train.py', '--config', str(CONFIG), '--data', str(DATA_PATH), '--output', str(OUTPUT)], check=True)
```

```python
import torch
checkpoint = torch.load(OUTPUT / 'best.pt', map_location='cpu', weights_only=False)
checkpoint['metrics']
```
